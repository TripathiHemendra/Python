```python
import numpy as np
arr = np.array([2,4,5,54,6,23])
print(np.sum(arr))
print(np.min(arr))
print(np.max(arr))
print(np.sum(arr)/len(arr))
print(np.cumsum(arr)) 
print(np.cumprod(arr)) #multiply elements
```

```python
# stactical fuction 
import numpy as np
import statistics as stas
arr = np.array([20,40,50,54,60,23,45,50]) 
print(np.mean(arr)) # avrage 
print(np.median(arr)) # first of all sort the iteam then find median value
print(stas.mode(arr)) # most frinquently number
print(np.std(arr))  # value spred means mean se value kitne kitne door pe hai
print(np.var(arr)) # square of standard deviation
```
